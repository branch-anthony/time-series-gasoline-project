# Setting-up Oil and Gas Data

```python
import pandas as pd

df = pd.read_csv("../../Data/Monday_Combined_dataset.csv")

df["date"] = pd.to_datetime(df["date"])
df = df.sort_values("date").set_index("date")

df = df.rename(columns={
    "gas_price": "gas_austin",
    "national_gas_price": "gas_us",
    "wti_price": "wti"
})

df.head()
```

# Granger Causality Check 

```python
from statsmodels.tsa.stattools import grangercausalitytests

# Use Monday-aligned dataset already loaded as df
# Columns should be: gas_austin, gas_us, wti

df_gc_austin = df[["gas_austin", "wti"]].diff().dropna()

print(df_gc_austin.head())
```

```python
print("WTI -> Austin Gas")
grangercausalitytests(df_gc_austin[["gas_austin", "wti"]], maxlag=4, verbose=True)

print("Austin Gas -> WTI")
grangercausalitytests(df_gc_austin[["wti", "gas_austin"]], maxlag=4, verbose=True)
```

```python
df_gc_us = df[["gas_us", "wti"]].diff().dropna()

print("WTI -> U.S. Gas")
grangercausalitytests(df_gc_us[["gas_us", "wti"]], maxlag=4, verbose=True)

print("U.S. Gas -> WTI")
grangercausalitytests(df_gc_us[["wti", "gas_us"]], maxlag=4, verbose=True)
```

The results show strong and consistent evidence that oil prices Granger-cause both Austin and U.S. gasoline prices across all lag lengths. The reverse direction is weaker and inconsistent for both markets. While there is some evidence that gasoline prices may Granger-cause oil prices at certain lags, this likely reflects shared market dynamics rather than a true causal relationship. Overall, both Austin and U.S. gasoline markets appear to respond to oil price movements rather than influence them.

# Cointegration test

```python
from statsmodels.tsa.stattools import coint

# Austin gas and WTI
coint_austin = coint(df["gas_austin"], df["wti"])

print("Austin Gas + WTI")
print("Test stat:", coint_austin[0])
print("p-value:", coint_austin[1])
print("Critical values:", coint_austin[2])  

print("\n")

# U.S. gas and WTI
coint_us = coint(df["gas_us"], df["wti"])

print("U.S. Gas + WTI")
print("Test stat:", coint_us[0])
print("p-value:", coint_us[1])
print("Critical values:", coint_us[2])
```

p = 0.01 < 0.05 → reject null

-> There IS cointegration

The cointegration test indicates a long-run equilibrium relationship between oil and gasoline prices. Combined with strong Granger causality from oil to gasoline prices, this suggests that a Vector Error Correction Model (VECM) is appropriate to capture both short-run dynamics and long-run equilibrium behavior.
